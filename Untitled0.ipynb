```python
!pip install rdkit pandas scikit-learn xgboost matplotlib shap -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from rdkit import Chem
from rdkit.Chem import Descriptors, rdFingerprintGenerator
from rdkit import DataStructs
from rdkit import RDLogger
RDLogger.DisableLog('rdApp.*')  # silence RDKit warnings

from sklearn.model_selection import (
    train_test_split, KFold, GridSearchCV, learning_curve, cross_val_score
)
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.inspection import permutation_importance
from sklearn.neighbors import NearestNeighbors
from xgboost import XGBRegressor
import shap
import joblib

# Global reproducibility
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print(f"✅ Imports complete. Random seed fixed at {RANDOM_STATE}.")
```

```python
url_co2 = "https://raw.githubusercontent.com/sptiwari/polymer_backbones_paper_data/main/Experimental_permeabilities_selectivities_and_SMILES_of_polymers/CO2_permeability_data.csv"
url_sel = "https://raw.githubusercontent.com/sptiwari/polymer_backbones_paper_data/main/Experimental_permeabilities_selectivities_and_SMILES_of_polymers/co2_n2_selectivity_data.csv"

df_co2_raw = pd.read_csv(url_co2)
df_sel_raw = pd.read_csv(url_sel)

# Strip whitespace from column names
df_co2_raw.columns = df_co2_raw.columns.str.strip()
df_sel_raw.columns = df_sel_raw.columns.str.strip()

print("=" * 70)
print("DATA LOADING AUDIT")
print("=" * 70)
print(f"CO2 file shape:       {df_co2_raw.shape}")
print(f"CO2 file columns:     {df_co2_raw.columns.tolist()}")
print(f"Selectivity file:     {df_sel_raw.shape}")
print(f"Selectivity columns:  {df_sel_raw.columns.tolist()}")
print()

df_co2 = df_co2_raw.rename(columns={
    'CO2 permeability (barrer)': 'CO2_perm'
})[['SMILES', 'CO2_perm']]

df_sel = df_sel_raw.rename(columns={
    'CO2/N2 Selectivity': 'CO2_N2_sel'
})[['SMILES', 'CO2_N2_sel']]

print(f"CO2 file cleaned:     {df_co2.shape}")
print(f"Selectivity file:     {df_sel.shape}")
```

```python
print("=" * 70)
print("DATA CLEANING AUDIT")
print("=" * 70)

n_start = len(df_co2)
print(f"\n[Step 1] CO2 file initial:                    {n_start} rows")

df = pd.merge(df_co2, df_sel, on='SMILES', how='inner')
print(f"[Step 2] After inner merge with selectivity:  {len(df)} rows "
      f"(dropped {n_start - len(df)} without selectivity)")

n_before = len(df)
df = df.dropna(subset=['SMILES', 'CO2_perm', 'CO2_N2_sel'])
print(f"[Step 3] After dropping NaN:                  {len(df)} rows "
      f"(dropped {n_before - len(df)})")

n_before = len(df)
df = df[(df['CO2_perm'] > 0) & (df['CO2_N2_sel'] > 0)]
print(f"[Step 4] After keeping positive values:       {len(df)} rows "
      f"(dropped {n_before - len(df)})")

def is_valid_smiles(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles.replace('*', 'C'))
        return mol is not None
    except:
        return False

n_before = len(df)
valid_mask = df['SMILES'].apply(is_valid_smiles)
df = df[valid_mask].copy()
print(f"[Step 5] After RDKit SMILES validation:       {len(df)} rows "
      f"(dropped {n_before - len(df)} invalid)")

n_before = len(df)
df = df.drop_duplicates(subset='SMILES', keep='first')
print(f"[Step 6] After removing duplicate SMILES:     {len(df)} rows "
      f"(dropped {n_before - len(df)})")

df['log_CO2_perm'] = np.log10(df['CO2_perm'])
df['log_CO2_N2_sel'] = np.log10(df['CO2_N2_sel'])

print(f"\nFinal dataset size: {len(df)} polymers")
print(f"CO2 permeability range: {df['CO2_perm'].min():.2f} – "
      f"{df['CO2_perm'].max():.2f} Barrer")
print(f"Selectivity range:      {df['CO2_N2_sel'].min():.2f} – "
      f"{df['CO2_N2_sel'].max():.2f}")
print("=" * 70)

df.to_csv('clean_data.csv', index=False)
print("✅ Saved: clean_data.csv")
```

```python
from rdkit.Chem import rdFingerprintGenerator

# Morgan generator — created ONCE, no deprecation warnings
MORGAN_GEN = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=128)

def cap_psmiles(smiles):
    return smiles.replace('*', 'C')

def generate_descriptors(smiles):
    try:
        mol = Chem.MolFromSmiles(cap_psmiles(smiles))
        if mol is None:
            return None
        return {
            'MolWt': Descriptors.MolWt(mol),
            'NumHeavyAtoms': Descriptors.HeavyAtomCount(mol),
            'NumRings': Descriptors.RingCount(mol),
            'NumAromaticRings': Descriptors.NumAromaticRings(mol),
            'NumHDonors': Descriptors.NumHDonors(mol),
            'NumHAcceptors': Descriptors.NumHAcceptors(mol),
            'TPSA': Descriptors.TPSA(mol),
            'MolLogP': Descriptors.MolLogP(mol),
            'NumRotatableBonds': Descriptors.NumRotatableBonds(mol),
            'FractionCSP3': Descriptors.FractionCSP3(mol),
            'NumValenceElectrons': Descriptors.NumValenceElectrons(mol),
            'NumHeteroatoms': Descriptors.NumHeteroatoms(mol),
            'MolMR': Descriptors.MolMR(mol),
            'LabuteASA': Descriptors.LabuteASA(mol),
            'BalabanJ': Descriptors.BalabanJ(mol),
            'BertzCT': Descriptors.BertzCT(mol),
            'NumSaturatedRings': Descriptors.NumSaturatedRings(mol),
            'NumAliphaticRings': Descriptors.NumAliphaticRings(mol),
            'NumAromaticHeterocycles': Descriptors.NumAromaticHeterocycles(mol),
            'NumSpiroAtoms': Descriptors.NumSpiroAtoms(mol),
        }
    except Exception:
        return None

def generate_morgan_fp(smiles):
    """Morgan fingerprint (radius=2, 128 bits) — modern RDKit API."""
    try:
        mol = Chem.MolFromSmiles(cap_psmiles(smiles))
        if mol is None:
            return None
        fp = MORGAN_GEN.GetFingerprint(mol)
        arr = np.zeros((128,), dtype=int)
        DataStructs.ConvertToNumpyArray(fp, arr)
        return arr
    except Exception:
        return None

# Build features
desc_rows, fp_rows, valid_idx = [], [], []
for idx, row in df.iterrows():
    d = generate_descriptors(row['SMILES'])
    f = generate_morgan_fp(row['SMILES'])
    if d is not None and f is not None:
        desc_rows.append(d)
        fp_rows.append(f)
        valid_idx.append(idx)

desc_df = pd.DataFrame(desc_rows, index=valid_idx)
fp_df = pd.DataFrame(fp_rows, index=valid_idx,
                     columns=[f'FP_{i}' for i in range(128)])

X_all = pd.concat([desc_df, fp_df], axis=1)
y_perm = df.loc[valid_idx, 'log_CO2_perm'].reset_index(drop=True)
y_sel  = df.loc[valid_idx, 'log_CO2_N2_sel'].reset_index(drop=True)
X_all  = X_all.reset_index(drop=True)

print("=" * 70)
print("FEATURE ENGINEERING AUDIT")
print("=" * 70)
print(f"Total features:       {X_all.shape[1]} (20 descriptors + 128 Morgan FP)")
print(f"Total samples:        {X_all.shape[0]}")
print(f"Feature/sample ratio: {X_all.shape[1]/X_all.shape[0]:.2f}")
print()

# Remove zero-variance features
variances = X_all.var()
zero_var = variances[variances == 0].index.tolist()
if zero_var:
    X_all = X_all.drop(columns=zero_var)
    print(f"[Step 1] Removed {len(zero_var)} zero-variance features.")
else:
    print(f"[Step 1] No zero-variance features found.")

# Remove highly correlated features
corr_matrix = X_all.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop = [col for col in upper.columns if any(upper[col] > 0.98)]
if to_drop:
    X_all = X_all.drop(columns=to_drop)
    print(f"[Step 2] Removed {len(to_drop)} highly correlated features (>0.98).")
else:
    print(f"[Step 2] No highly correlated features found.")

print(f"\nFinal feature matrix: {X_all.shape}")
print("=" * 70)
```

```python
print("=" * 70)
print("SPLIT STRATEGY")
print("=" * 70)

X_temp, X_test, y_perm_temp, y_perm_test, y_sel_temp, y_sel_test = train_test_split(
    X_all, y_perm, y_sel, test_size=0.15, random_state=RANDOM_STATE
)
X_train, X_val, y_perm_train, y_perm_val, y_sel_train, y_sel_val = train_test_split(
    X_temp, y_perm_temp, y_sel_temp, test_size=0.176, random_state=RANDOM_STATE
)

print(f"Train:      {X_train.shape[0]} samples ({X_train.shape[0]/len(X_all)*100:.1f}%)")
print(f"Validation: {X_val.shape[0]} samples ({X_val.shape[0]/len(X_all)*100:.1f}%)")
print(f"Test:       {X_test.shape[0]} samples ({X_test.shape[0]/len(X_all)*100:.1f}%)")
print(f"\nTest set is NEVER used for tuning.")
print("=" * 70)
```

```python
print("=" * 70)
print("BASELINE MODELS")
print("=" * 70)

# ─── Baseline 1: Mean predictor ───
y_perm_mean = y_perm_train.mean()
y_sel_mean = y_sel_train.mean()
perm_mean_pred = np.full_like(y_perm_test, y_perm_mean)
sel_mean_pred = np.full_like(y_sel_test, y_sel_mean)

print(f"\n[Baseline 1] Mean predictor:")
print(f"  Permeability: R² = {r2_score(y_perm_test, perm_mean_pred):.4f}  "
      f"MAE = {mean_absolute_error(y_perm_test, perm_mean_pred):.4f}")
print(f"  Selectivity:  R² = {r2_score(y_sel_test, sel_mean_pred):.4f}  "
      f"MAE = {mean_absolute_error(y_sel_test, sel_mean_pred):.4f}")

# ─── Baseline 2: Linear Regression ───
scaler_lr = StandardScaler()
X_train_scaled = scaler_lr.fit_transform(X_train)
X_val_scaled   = scaler_lr.transform(X_val)
X_test_scaled  = scaler_lr.transform(X_test)

lr_perm = LinearRegression().fit(X_train_scaled, y_perm_train)
lr_sel  = LinearRegression().fit(X_train_scaled, y_sel_train)

lr_perm_test = lr_perm.predict(X_test_scaled)
lr_sel_test  = lr_sel.predict(X_test_scaled)

print(f"\n[Baseline 2] Linear Regression:")
print(f"  Permeability — Val R²:  {r2_score(y_perm_val, lr_perm.predict(X_val_scaled)):.4f}")
print(f"  Permeability — Test R²: {r2_score(y_perm_test, lr_perm_test):.4f}")
print(f"  Selectivity  — Val R²:  {r2_score(y_sel_val, lr_sel.predict(X_val_scaled)):.4f}")
print(f"  Selectivity  — Test R²: {r2_score(y_sel_test, lr_sel_test):.4f}")

# ─── Baseline 3: Ridge ───
ridge_perm = Ridge(alpha=1.0).fit(X_train_scaled, y_perm_train)
ridge_sel  = Ridge(alpha=1.0).fit(X_train_scaled, y_sel_train)

print(f"\n[Baseline 3] Ridge Regression:")
print(f"  Permeability — Val R²:  {r2_score(y_perm_val, ridge_perm.predict(X_val_scaled)):.4f}")
print(f"  Permeability — Test R²: {r2_score(y_perm_test, ridge_perm.predict(X_test_scaled)):.4f}")
print(f"  Selectivity  — Val R²:  {r2_score(y_sel_val, ridge_sel.predict(X_val_scaled)):.4f}")
print(f"  Selectivity  — Test R²: {r2_score(y_sel_test, ridge_sel.predict(X_test_scaled)):.4f}")
print("=" * 70)
```

```python
print("=" * 70)
print("HYPERPARAMETER TUNING (train+val, 5-fold CV)")
print("=" * 70)

X_tune = pd.concat([X_train, X_val], axis=0)
y_perm_tune = pd.concat([y_perm_train, y_perm_val], axis=0)
y_sel_tune  = pd.concat([y_sel_train,  y_sel_val], axis=0)

# Random Forest grid
rf_grid = {
    'n_estimators': [300, 500],
    'max_depth': [6, 10, 15],
    'min_samples_leaf': [1, 3, 5],
    'max_features': ['sqrt', 0.3],
}

rf_gs_perm = GridSearchCV(
    RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=-1),
    rf_grid, cv=5, scoring='r2', n_jobs=-1)
rf_gs_perm.fit(X_tune, y_perm_tune)
print(f"RF (perm) best params: {rf_gs_perm.best_params_}")
print(f"RF (perm) best CV R²:  {rf_gs_perm.best_score_:.4f}")

rf_gs_sel = GridSearchCV(
    RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=-1),
    rf_grid, cv=5, scoring='r2', n_jobs=-1)
rf_gs_sel.fit(X_tune, y_sel_tune)
print(f"RF (sel) best params:  {rf_gs_sel.best_params_}")
print(f"RF (sel) best CV R²:   {rf_gs_sel.best_score_:.4f}")

# XGBoost grid
xgb_grid = {
    'n_estimators': [300, 500],
    'max_depth': [4, 6, 8],
    'learning_rate': [0.03, 0.05, 0.1],
    'subsample': [0.7, 0.9],
}

xgb_gs_perm = GridSearchCV(
    XGBRegressor(random_state=RANDOM_STATE, n_jobs=-1),
    xgb_grid, cv=5, scoring='r2', n_jobs=-1)
xgb_gs_perm.fit(X_tune, y_perm_tune)
print(f"\nXGB (perm) best params: {xgb_gs_perm.best_params_}")
print(f"XGB (perm) best CV R²:  {xgb_gs_perm.best_score_:.4f}")

xgb_gs_sel = GridSearchCV(
    XGBRegressor(random_state=RANDOM_STATE, n_jobs=-1),
    xgb_grid, cv=5, scoring='r2', n_jobs=-1)
xgb_gs_sel.fit(X_tune, y_sel_tune)
print(f"XGB (sel) best params:  {xgb_gs_sel.best_params_}")
print(f"XGB (sel) best CV R²:   {xgb_gs_sel.best_score_:.4f}")
print("=" * 70)
```

```python
# Final RF models
rf_perm = RandomForestRegressor(**rf_gs_perm.best_params_,
                                 random_state=RANDOM_STATE, n_jobs=-1)
rf_perm.fit(X_tune, y_perm_tune)

rf_sel = RandomForestRegressor(**rf_gs_sel.best_params_,
                                random_state=RANDOM_STATE, n_jobs=-1)
rf_sel.fit(X_tune, y_sel_tune)

# Final XGBoost models
xgb_perm = XGBRegressor(**xgb_gs_perm.best_params_,
                         random_state=RANDOM_STATE, n_jobs=-1)
xgb_perm.fit(X_tune, y_perm_tune)

xgb_sel = XGBRegressor(**xgb_gs_sel.best_params_,
                        random_state=RANDOM_STATE, n_jobs=-1)
xgb_sel.fit(X_tune, y_sel_tune)

print("✅ Production models trained:")
print("   - Random Forest (perm + sel)")
print("   - XGBoost (perm + sel)")
print("\n❌ Neural Network dropped — insufficient data for deep learning")
```

```python
print("=" * 70)
print("5-FOLD CROSS-VALIDATION (train+val set)")
print("=" * 70)

cv_rf_perm  = cross_val_score(rf_perm,  X_tune, y_perm_tune, cv=5, scoring='r2')
cv_rf_sel   = cross_val_score(rf_sel,   X_tune, y_sel_tune,  cv=5, scoring='r2')
cv_xgb_perm = cross_val_score(xgb_perm, X_tune, y_perm_tune, cv=5, scoring='r2')
cv_xgb_sel  = cross_val_score(xgb_sel,  X_tune, y_sel_tune,  cv=5, scoring='r2')

def fmt(s): return f"{s.mean():.4f} ± {s.std():.4f}"

print(f"\n{'Model':<12} {'Target':<15} {'CV R²':<22}")
print("-" * 50)
print(f"{'RF':<12} {'Permeability':<15} {fmt(cv_rf_perm):<22}")
print(f"{'RF':<12} {'Selectivity':<15} {fmt(cv_rf_sel):<22}")
print(f"{'XGBoost':<12} {'Permeability':<15} {fmt(cv_xgb_perm):<22}")
print(f"{'XGBoost':<12} {'Selectivity':<15} {fmt(cv_xgb_sel):<22}")
print("=" * 70)
```

```python
def bootstrap_ci(y_true, y_pred, n_boot=1000, ci=95):
    r2s = []
    n = len(y_true)
    for _ in range(n_boot):
        idx = np.random.randint(0, n, n)
        r2s.append(r2_score(y_true.iloc[idx], y_pred[idx]))
    lo = np.percentile(r2s, (100 - ci) / 2)
    hi = np.percentile(r2s, 100 - (100 - ci) / 2)
    return lo, hi

predictions = {
    'Linear':  {'perm': lr_perm.predict(X_test_scaled),
                'sel':  lr_sel.predict(X_test_scaled)},
    'RF':      {'perm': rf_perm.predict(X_test),
                'sel':  rf_sel.predict(X_test)},
    'XGBoost': {'perm': xgb_perm.predict(X_test),
                'sel':  xgb_sel.predict(X_test)},
}

print("=" * 95)
print("TEST SET EVALUATION (95% bootstrap CIs)")
print("=" * 95)
print(f"{'Model':<10} {'Target':<14} {'R²':<9} {'95% CI':<22} {'MAE':<10} {'RMSE':<10}")
print("-" * 95)

results_rows = []
for model_name, preds in predictions.items():
    for target_name, y_true in [('Permeability', y_perm_test),
                                 ('Selectivity', y_sel_test)]:
        y_pred = preds['perm'] if target_name == 'Permeability' else preds['sel']
        r2   = r2_score(y_true, y_pred)
        mae  = mean_absolute_error(y_true, y_pred)
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))
        lo, hi = bootstrap_ci(y_true, y_pred)
        print(f"{model_name:<10} {target_name:<14} {r2:<9.4f} "
              f"[{lo:.3f}, {hi:.3f}]      {mae:<10.4f} {rmse:<10.4f}")
        results_rows.append({
            'Model': model_name, 'Target': target_name,
            'R2': r2, 'R2_CI_low': lo, 'R2_CI_high': hi,
            'MAE': mae, 'RMSE': rmse
        })

results_df = pd.DataFrame(results_rows)
results_df.to_csv('model_results.csv', index=False)
print("=" * 95)
print("✅ Saved: model_results.csv")
```

```python
nbrs = NearestNeighbors(n_neighbors=5).fit(X_tune)

# Compute training distribution
train_dists, _ = nbrs.kneighbors(X_tune)
train_mean_dists = train_dists[:, 1:].mean(axis=1)

p50 = np.percentile(train_mean_dists, 50)
p90 = np.percentile(train_mean_dists, 90)
p95 = np.percentile(train_mean_dists, 95)

print("=" * 70)
print("APPLICABILITY DOMAIN — Reference Distribution")
print("=" * 70)
print(f"Training mean-neighbor distance:")
print(f"  Median (P50): {p50:.3f}")
print(f"  P90:          {p90:.3f}")
print(f"  P95:          {p95:.3f}")
print(f"  Max:          {train_mean_dists.max():.3f}")
print("=" * 70)

def applicability_domain(smiles):
    d = generate_descriptors(smiles)
    f = generate_morgan_fp(smiles)
    if d is None or f is None:
        return None
    X_new = pd.DataFrame([{**d, **{f'FP_{i}': f[i] for i in range(128)}}])
    X_new = X_new[X_all.columns]
    dist, _ = nbrs.kneighbors(X_new.values)
    mean_dist = dist[:, 1:].mean()

    if mean_dist <= p50:
        rel = "High (in-domain)"
    elif mean_dist <= p90:
        rel = "Moderate (in-domain)"
    elif mean_dist <= p95:
        rel = "Low-moderate (near boundary)"
    else:
        rel = "Low (out-of-domain)"
    return mean_dist, rel

print("\nApplicability domain check (first 5 test polymers):")
for i, smiles in enumerate(df.loc[valid_idx[:5], 'SMILES']):
    result = applicability_domain(smiles)
    if result:
        dist, rel = result
        print(f"  Polymer {i+1}: distance = {dist:.3f} → {rel}")
```

```python
def predict_polymer(smiles_string):
    """Predict using best model per target."""
    d = generate_descriptors(smiles_string)
    f = generate_morgan_fp(smiles_string)
    if d is None or f is None:
        return "Invalid SMILES"

    X_new = pd.DataFrame([{**d, **{f'FP_{i}': f[i] for i in range(128)}}])
    X_new = X_new[X_all.columns]

    perm_model = xgb_perm if best_perm_name == 'XGBoost' else rf_perm
    sel_model  = rf_sel   if best_sel_name  == 'RF'      else xgb_sel

    log_perm = perm_model.predict(X_new)[0]
    log_sel  = sel_model.predict(X_new)[0]

    dist, rel = applicability_domain(smiles_string)

    return {
        'SMILES': smiles_string,
        'CO2_Permeability_Barrer': round(10 ** log_perm, 2),
        'CO2_N2_Selectivity': round(10 ** log_sel, 2),
        'Applicability': rel,
    }

sample_smiles = df.iloc[0]['SMILES']
print("=" * 70)
print("PREDICTION DEMO")
print("=" * 70)
print(f"\nSMILES: {sample_smiles}")
print(f"Actual:    P = {df.iloc[0]['CO2_perm']:.2f} Barrer, "
      f"α = {df.iloc[0]['CO2_N2_sel']:.2f}")
print(f"Predicted: {predict_polymer(sample_smiles)}")
print("=" * 70)
```
